```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('creditcard.csv')
print(df.info())

print(df['Class'].value_counts())
print(df['Class'].value_counts(normalize=True)* 100)
print(df.groupby('Class')['Amount'].describe())
```

## Explain the dataset

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.boxplot(x='Class', y='Amount', data = df, ax=axes[0])
axes[0].set_ylim(0,500)
axes[0].set_title('Boxplot of Amount by Class')
sns.histplot(data=df, x='Time', hue='Class', bins=50, stat='density', common_norm=False, ax=axes[1])
axes[1].set_title('Histogram of Time by Class')
plt.tight_layout()
plt.show()
```

```python
# heatmap of correlation matrix
corr = df.corr()['Class'].drop('Class')
colors = ['#c0392b' if c < 0 else '#2980b9' for c in corr]

fig, ax = plt.subplots(figsize=(6, 6))
ax.grid(axis='x', linestyle='--', alpha=0.3)
ax.barh(corr.index, corr.values, color=colors)
ax.axvline(0, color='black', linewidth=0.8)
ax.set_xlabel('Correlation with Class')
ax.set_ylabel('Feature Correlation with Fraud Label', fontsize=13, fontweight='bold')
ax.spines[['top', 'right']].set_visible(False)
plt.tight_layout()
plt.show()
```
